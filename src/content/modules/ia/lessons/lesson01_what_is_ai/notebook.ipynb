# Lección 01: ¿Qué es la Inteligencia Artificial?

Exploramos el dataset Breast Cancer Wisconsin para entender cómo las **features** (características numéricas) permiten a la IA reconocer patrones y clasificar datos.

## Setup

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer

plt.style.use('seaborn-v0_8-whitegrid')
np.set_printoptions(precision=2, suppress=True)
```

## Cargar el dataset Breast Cancer Wisconsin

Este dataset contiene 569 casos reales de biopsias. Cada caso tiene 30 features numéricas extraídas de imágenes de núcleos celulares, y está etiquetado como **maligno** o **benigno**.

```python
data = load_breast_cancer()
X = data.data
y = data.target
feature_names = data.feature_names
target_names = data.target_names

print(f"Número de muestras: {X.shape[0]}")
print(f"Número de features: {X.shape[1]}")
print(f"Clases: {target_names}")
```

## Explorar features

Veamos la forma de los datos, los nombres de las features y la distribución de las clases.

```python
print(f"Shape de X (muestras × features): {X.shape}")
print(f"Shape de y (labels): {y.shape}")
print(f"\nNombres de las features:")
for i, name in enumerate(feature_names):
    print(f"  {i+1:2d}. {name}")
```

### Distribución de las clases

```python
unique, counts = np.unique(y, return_counts=True)

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(target_names, counts, color=['#e74c3c', '#2ecc71'])
ax.set_title('Distribución de clases')
ax.set_ylabel('Cantidad de muestras')

for i, (name, count) in enumerate(zip(target_names, counts)):
    ax.text(i, count + 5, f'{count} ({count/len(y)*100:.1f}%)', ha='center')

plt.tight_layout()
plt.show()
```

## Visualización 2D: Scatter plot de features

Visualizamos cómo se distribuyen las muestras en el espacio de features. Cada punto es un caso real de biopsia.

```python
feat_x = 0  # mean radius
feat_y = 2  # mean perimeter

malignant = y == 0
benign = y == 1

fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(X[malignant, feat_x], X[malignant, feat_y],
           c='#e74c3c', label='Malignant', alpha=0.6, edgecolors='k', s=40)
ax.scatter(X[benign, feat_x], X[benign, feat_y],
           c='#2ecc71', label='Benign', alpha=0.6, edgecolors='k', s=40)

ax.set_xlabel(feature_names[feat_x])
ax.set_ylabel(feature_names[feat_y])
ax.set_title(f'{feature_names[feat_x]} vs {feature_names[feat_y]}')
ax.legend()

plt.tight_layout()
plt.show()
```

## Comparación de features: Benigno vs Maligno

Calculamos estadísticas por clase para identificar las features más discriminativas.

```python
malignant_data = X[y == 0]
benign_data = X[y == 1]

print(f"{'Feature':<25} {'Malignant':>12} {'Benign':>12} {'Diff':>10}")
print("-" * 62)

differences = []
for i, name in enumerate(feature_names[:10]):
    mal_mean = malignant_data[:, i].mean()
    ben_mean = benign_data[:, i].mean()
    diff = mal_mean - ben_mean
    rel_diff = abs(diff / ben_mean) * 100
    differences.append((name, rel_diff))
    print(f"{name:<25} {mal_mean:>12.2f} {ben_mean:>12.2f} {diff:>10.2f}")

differences.sort(key=lambda x: x[1], reverse=True)
print(f"\nTop 3 features con mayor diferencia relativa:")
for name, pct in differences[:3]:
    print(f"  {name}: {pct:.1f}%")
```

### Boxplot comparativo de las features más discriminativas

```python
top_features = [d[0] for d in differences[:5]]

fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for idx, fname in enumerate(top_features):
    feat_idx = list(feature_names).index(fname)
    axes[idx].boxplot(
        [X[y == 0, feat_idx], X[y == 1, feat_idx]],
        labels=['Malignant', 'Benign']
    )
    axes[idx].set_title(fname, fontsize=9, rotation=15)
    axes[idx].tick_params(labelsize=8)

plt.suptitle('Top 5 features más discriminativas', fontsize=12)
plt.tight_layout()
plt.show()
```

## Resumen

- El dataset Breast Cancer Wisconsin tiene **569 muestras** con **30 features** numéricas
- Las clases son **malignant** (212) y **benign** (357) — no están perfectamente balanceadas
- Algunas features como **mean radius**, **mean perimeter** y **mean area** muestran diferencias claras entre clases
- Sin embargo, siempre hay **superposición** entre las clases — por eso las reglas fivas no alcanzan y necesitamos machine learning
- Las features son el **idioma de la máquina**: convierten propiedades del mundo real en números que la IA puede procesar
