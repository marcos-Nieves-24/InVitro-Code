# Lección 3: Variables

Aprende a almacenar y gestionar datos usando variables en Python.

## Asignación básica de variables

```python
import math  # noqa: F401

# Assign values to variables
gene = "BRCA1"
expression_level = 2.5
is_mutated = True

print("Gene:", gene)
print("Expression:", expression_level)
print("Mutated:", is_mutated)
```

## Tipado dinámico

```python
# Python infers the type from the value
x = 42
print(f"x = {x}, type = {type(x)}")

x = 3.14159
print(f"x = {x}, type = {type(x)}")

x = "Hello"
print(f"x = {x}, type = {type(x)}")

x = [1, 2, 3]
print(f"x = {x}, type = {type(x)}")
```

## Asignación múltiple

```python
# Tuple unpacking
a, b, c = 10, 20, 30
print(f"a={a}, b={b}, c={c}")

# Same value to multiple variables
x = y = z = 0
print(f"x={x}, y={y}, z={z}")

# Swapping
a, b = b, a
print(f"After swap: a={a}, b={b}")
```

## Entrada del usuario

```python
# Simulating user input with hardcoded values
name = "Alice"
age = 30
height = 1.75

print(f"Hello {name}, you are {age} years old and {height}m tall.")
```

## Ejemplo de biotecnología: Registros de pacientes

```python
patient_id = "P-0042"
age = 45
heart_rate = 72
systolic_bp = 120
diastolic_bp = 80
diagnosis = "Hypertension"

print(f"Patient: {patient_id}")
print(f"Age: {age} years")
print(f"Heart Rate: {heart_rate} bpm")
print(f"Blood Pressure: {systolic_bp}/{diastolic_bp} mmHg")
print(f"Diagnosis: {diagnosis}")
```

## Ejemplo de SaaS: Métricas de negocio

```python
company = "DataCloud Inc."
revenue = 125000.50
users = 15420
churn_rate = 0.035
arpu = revenue / users

print(f"Company: {company}")
print(f"Monthly Revenue: ${revenue:,.2f}")
print(f"Active Users: {users:,}")
print(f"Churn Rate: {churn_rate:.1%}")
print(f"ARPU: ${arpu:.2f}")
```

## Resumen

- Las variables almacenan datos en ubicaciones de memoria con nombre
- Python es de tipado dinámico: el tipo se infiere del valor
- Usa nombres en `snake_case` (PEP 8)
- `input()` siempre devuelve una cadena de texto
- La asignación múltiple y el desempaquetado de tuplas son características potentes de Python
